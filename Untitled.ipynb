```python
import skcriteria as skc
from skcriteria.agg.topsis import TOPSIS
from skcriteria.pipelines import mkpipe
from skcriteria.preprocessing.invert_objectives import InvertMinimize
from skcriteria.preprocessing.scalers import SumScaler
from skcriteria.preprocessing.weighters import EntropyWeighter
from skcriteria.ranksrev.rank_transitivity_check import RankTransitivityChecker

pipe = mkpipe(
    InvertMinimize(),
    EntropyWeighter(),
    SumScaler(target="matrix"),
    TOPSIS(),
)

checker = RankTransitivityChecker(pipe, allow_missing_alternatives=True)

dm = skc.datasets.load_van2021evaluation()

result = checker.evaluate(dm)
```

```python
G = result.extra_.pairwise_dominance_graph
```

```python
import matplotlib.pyplot as plt
import networkx as nx

def plot(G):

# 3. Configure the drawing with visible labels
    nx.draw(G, with_labels=True, node_color="skyblue", edge_color="gray", node_size=700)

# 4. Display the window containing the rendering
    plt.show()

plot(G)

nx.is_directed_acyclic_graph(G)
```

```python
 
import networkx as nx
import igraph as ig
import numpy as np

def resolve_fas_method(graph, method):
    if method == "auto":
        method = "ip" if graph.number_of_nodes() < 100 else "eades"
    return method
 
 
def _nx_fas(igraph_g, fas):
    nx_edges = []
    for edge in igraph_g.es[fas]:
        nx_edge = tuple(
            node.attributes()["_nx_name"] for node in edge.vertex_tuple
        )
        nx_edges.append(nx_edge)
    return nx_edges
 
 
def as_reduced_dag(graph, *, method="auto"):
    """Convert a directed graph into a reduced, acyclic DAG.
 
    Breaks cycles using the minimum Feedback Arc Set (FAS), then applies
    transitive reduction on the resulting DAG to remove edges implied by
    other paths (e.g. if A->B and B->C both exist, a direct A->C edge is
    redundant and gets dropped).
 
    Typical input: tournament-like graphs (e.g. a pairwise dominance
    graph, where every pair of nodes has a directed edge), which tend to
    be dense and contain many transitively redundant edges. Neither step
    changes the results of downstream ranking reconstruction (see the
    "invariancia de toposorts" test below).
    """
    dag, fas, rmethod = graph, [], None
    if not nx.is_directed_acyclic_graph(graph):
        rmethod = resolve_fas_method(graph, method)
        igraph_g = ig.Graph.from_networkx(graph)
        fas_ids = igraph_g.feedback_arc_set(method=method)
        fas = _nx_fas(igraph_g, fas_ids)
     
        igraph_g.delete_edges(fas_ids)
        dag = igraph_g.to_networkx()
    dag = nx.transitive_reduction(dag)
 
    return dag, fas, rmethod
```
